```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, roc_auc_score
```

```python
import joblib

# قراءة البيانات الجاهزة فوراً
X_train_scaled = joblib.load('D:/ML Projects/E-Commerce Customer Churn/data/pkl/X_train_scaled.pkl')
X_test_scaled = joblib.load('D:/ML Projects/E-Commerce Customer Churn/data/pkl/X_test_scaled.pkl')
y_train = joblib.load('D:/ML Projects/E-Commerce Customer Churn/data/pkl/y_train.pkl')
y_test = joblib.load('D:/ML Projects/E-Commerce Customer Churn/data/pkl/y_test.pkl')
```

```python
df=pd.read_csv("D:/ML Projects/E-Commerce Customer Churn/data/processed/prepared_featureEngineering.csv")
```

```python
df
```

```python
top_8_features =[
    'days_since_last_purchase',
    'engagement_score',
    'customer_support_tickets',
    'satisfaction_score',
    'loyalty_member',
    'inactivity_risk_score',
    'product_review_score_avg',
    'discount_usage_rate'
]
```

```python
X_top8 = df[top_8_features]
y = df['churned']
```

```python
X_train_8, X_test_8, y_train, y_test = train_test_split(
    X_top8, y, test_size=0.2, random_state=42, stratify=y
)
```

```python
scaler_8 = StandardScaler()
X_train_8_scaled = scaler_8.fit_transform(X_train_8)
X_test_8_scaled = scaler_8.transform(X_test_8)
```

```python
log_reg_8 = LogisticRegression(random_state=42)
log_reg_8.fit(X_train_8_scaled, y_train)

y_pred_lr = log_reg_8.predict(X_test_8_scaled)
y_prob_lr = log_reg_8.predict_proba(X_test_8_scaled)[:, 1]

print("=== Logistic Regression (Top 8 Features) ===")
print(f"Accuracy : {accuracy_score(y_test, y_pred_lr):.4f}")
print(f"ROC-AUC  : {roc_auc_score(y_test, y_prob_lr):.4f}")
print("\nClassification Report:\n", classification_report(y_test, y_pred_lr))
```

```python
rf_8 = RandomForestClassifier(random_state=42)
rf_8.fit(X_train_8_scaled, y_train)

y_pred_rf = rf_8.predict(X_test_8_scaled)
y_prob_rf = rf_8.predict_proba(X_test_8_scaled)[:, 1]

print("\n" + "="*50 + "\n")
print("=== Random Forest (Top 8 Features) ===")
print(f"Accuracy : {accuracy_score(y_test, y_pred_rf):.4f}")
print(f"ROC-AUC  : {roc_auc_score(y_test, y_prob_rf):.4f}")
print("\nClassification Report:\n", classification_report(y_test, y_pred_rf))
```

```python
from sklearn.model_selection import GridSearchCV

# 1. تحديد الـ Hyperparameters المراد ضبطها
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'class_weight': [None, 'balanced'] # مفيد جداً للبيانات غير المتوازنة لرفع الـ Recall
}

# 2. إنشاء الـ GridSearchCV مع التركيز على ROC-AUC أو Recall
rf_base = RandomForestClassifier(random_state=42)

grid_search = GridSearchCV(
    estimator=rf_base,
    param_grid=param_grid,
    scoring='roc_auc',  # أو 'f1' / 'recall'
    cv=5,
    n_jobs=-1,
    verbose=1
)

# 3. التشغيل على بيانات الـ Top 8
grid_search.fit(X_train_8_scaled, y_train)

# 4. عرض أفضل المعاملات وأفضل أداء أثناء الـ CV
print("=== Best Parameters Found ===")
print(grid_search.best_params_)
print(f"\nBest Cross-Validation ROC-AUC: {grid_search.best_score_:.4f}")

# 5. تقييم الموديل الأفضل (Best Estimator) على الـ Test Set
best_rf = grid_search.best_estimator_
y_pred_best = best_rf.predict(X_test_8_scaled)
y_prob_best = best_rf.predict_proba(X_test_8_scaled)[:, 1]

print("\n=== Tuned Random Forest Test Results ===")
print(f"Accuracy : {accuracy_score(y_test, y_pred_best):.4f}")
print(f"ROC-AUC  : {roc_auc_score(y_test, y_prob_best):.4f}")
print("\nClassification Report:\n", classification_report(y_test, y_pred_best))
```
