```python
import numpy as np 
import pandas as pd
import sklearn 
from sklearn.model_selection import train_test_split
#import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("D:/ML Projects/E-Commerce Customer Churn/data/processed/Encoded_customer_data.csv")
```

Feature engineering-->add new features

```python
df['Orders_per_Month']=df['total_orders']/df['account_age_months']+ 1
df['total_customer_spend'] = df['avg_order_value'] * df['total_orders']
df['inactivity_risk_score'] = df['days_since_last_purchase'] / (
    df['browsing_frequency_per_week'] + 1)
df['frustration_index'] = (
    df['customer_support_tickets'] * df['return_rate']
) / (df['satisfaction_score'] + 1)
df['discount_hunter'] = (
    df['discount_usage_rate'] * df['price_sensitivity_index']
)
```

```python
df.head()
```

 Train-Test Split

```python
y=df['churned']
x=df.drop(columns=['churned'])
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.20,random_state=42,stratify=y)
```

```python
print("حجم بيانات التدريب:", xtrain.shape)
print("حجم بيانات الاختبار:", xtest.shape)
```

Encoding & Scaling

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
xtrain_scaled=scaler.fit_transform(xtrain)
xtest_scaled=scaler.transform(xtest)
```

Wrapper selection ممكن نبدا ب forward selection

```python
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.linear_model import LogisticRegression 
```

```python
base_model = LogisticRegression(class_weight='balanced', random_state=42)
```

```python
forward = SequentialFeatureSelector(
    estimator=base_model,
    n_features_to_select=10,
    direction='forward',  # تحديد الإتجاه الأمامي
    scoring='roc_auc',  # معيار التقييم (ممتاز للـ Churn)
    cv=5,  # استخدام 5-Fold Cross Validation لتفادي overfitting
    n_jobs=-1,  # لتسريع العملية باستخدام كل أنوية المعالج
)
```

```python
forward.fit(xtrain_scaled, ytrain)
```

```python
selected_forward = xtrain.columns[forward.get_support()]
print(list(selected_forward))
```
